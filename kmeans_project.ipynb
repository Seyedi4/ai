```python
import random
import math
```

```python
data = [[1, 2], [5, 5], [10, 10], [12, 12]]
k=3
```

```python
def fasele(p1, p2):
    return math.sqrt((p1[0] - p2[0])**2 + (p1[1] - p2[1])**2)
```

```python
def nazdiktarin(p, marakez):
    fasele_ha = []
    for m in marakez:
        fasele_ha.append(fasele(p, m))

    min_fasele = min(fasele_ha)
    return fasele_ha.index(min_fasele)
```

```python
def mianghin(khushe):
    sum_x = 0
    sum_y = 0
    for p in khushe:
        sum_x = sum_x + p[0]
        sum_y = sum_y + p[1]
    return [sum_x / len(khushe), sum_y / len(khushe)]
```

```python
def marakez_jadid(khushe_ha, marakez_ghadimi, k):
    list_jadid = []
    for i in range(k):
        if len(khushe_ha[i]) > 0:
            markaz_jadid = mianghin(khushe_ha[i])
            list_jadid.append(markaz_jadid)
        else:
            list_jadid.append(marakez_ghadimi[i])
    return list_jadid
```

```python
def k_means(data, k):
    marakez = random.sample(data, k)
    
    while True:
        khushe_ha = []
        for i in range(k):
            khushe_ha.append([])
        
        for p in data:
            a = nazdiktarin(p, marakez)
            khushe_ha[a].append(p)
        
        marakez_ghadimi = list(marakez)
        
        marakez = marakez_jadid(khushe_ha, marakez_ghadimi, k)
        
        if marakez_ghadimi == marakez:
            break
            
    return marakez, khushe_ha
```

```python
marakez, clusters = k_means(data, k)
print("نقطه های ما یا (مراکز):", marakez)
print("--------------------------")
print("خوشه بندی یا دسته بندی:", clusters)
```
